```python
import geopandas as gpd

def get_label_column(gdf):
    for col in ["Cover", "Field", "Category"]:
        if col in gdf.columns:
            return col
    raise ValueError(f"No label column found. Columns: {gdf.columns.tolist()}")

def normalize_label(x):
    if x is None:
        return None

    x = str(x).lower().strip()

    if "sub" in x:
        return "submergent"
    elif "em" in x:
        return "emergent"
    elif "soil" in x:
        return "loamy soil"
    else:
        return x
```

```python
all_labels = set()

for shp in shp_files:
    gdf = gpd.read_file(shp)

    label_col = get_label_column(gdf)

    # Create a NEW standardized column
    gdf["label"] = gdf[label_col].apply(normalize_label)

    print(f"\n{shp}")
    print(gdf["label"].value_counts())

    all_labels.update(gdf["label"].unique())

print("\nAll labels found:")
```
